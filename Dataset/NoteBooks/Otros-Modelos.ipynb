```python
# Importar las librerías necesarias
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, f1_score

df4 = pd.read_csv(r"C:\\Users\\crist\\Desktop\\CursoEOI-IA\\Chapter-5\\Proyecto_Samsung\\Dataset\\dataset_bullying_final_2.csv")

# Dividir los datos en Train/Test con estratificación
X = df4['text']
y = df4['label']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# Crear pipelines para cada modelo
pipelines = {
    'Naive Bayes': Pipeline([
        ('tfidf', TfidfVectorizer(stop_words='english')),
        ('clf', MultinomialNB())
    ]),
    'LinearSVC': Pipeline([
        ('tfidf', TfidfVectorizer(stop_words='english')),
        ('clf', LinearSVC(random_state=42, class_weight='balanced', dual='auto'))
    ]),
    'Random Forest': Pipeline([
        ('tfidf', TfidfVectorizer(stop_words='english')),
        ('clf', RandomForestClassifier(random_state=42, class_weight='balanced', n_jobs=-1))
    ])
}

# Entrenar y evaluar
print("--- COMPARATIVA CON BALANCEO DE CLASES ---\n")

for model_name, pipeline in pipelines.items():
    print(f"modelo: {model_name}")
    pipeline.fit(X_train, y_train)
    y_pred = pipeline.predict(X_test)
    
    # Imprimimos reporte completo para ver el Recall de la clase 0
    print(classification_report(y_test, y_pred, target_names=['No Bullying', 'Bullying']))
    
    # F1 Score de la clase positiva
    f1 = f1_score(y_test, y_pred)
    print(f"👉 F1-score (Clase Bullying): {f1:.4f}\n")
    print("-" * 30)
```
