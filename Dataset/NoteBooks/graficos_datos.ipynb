```python
# 1. Importar las librerías necesarias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

```python
# Gráfico de queso (pie chart) con porcentajes
datos = [57815, 39489]
etiquetas = ['not_bullying', 'bullying']
colores = ['#3498db', '#e74c3c']

plt.figure(figsize=(8, 8))
plt.pie(datos, labels=etiquetas, autopct='%1.1f%%', colors=colores, startangle=90,
        explode=(0.02, 0.02), shadow=True, textprops={'fontsize': 12})
plt.title(f'Graphic distribution\n(Total: {sum(datos):,} registers)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Cyberbullying types data
categories = ['religion', 'age', 'gender', 'ethnicity', 'other_cyberbullying']
values = [7998, 7992, 7973, 7961, 7823]
total = sum(values)
percentages = [v / total * 100 for v in values]

# Elegant style configuration
sns.set_style("whitegrid")
colors = sns.color_palette("husl", len(categories))

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Chart 1: Horizontal bar chart with percentages
ax1 = axes[0, 0]
bars = sns.barplot(x=values, y=categories, palette=colors, ax=ax1)
ax1.set_xlabel('Count', fontsize=12)
ax1.set_ylabel('Category', fontsize=12)
ax1.set_title('Distribution by Cyberbullying Type', fontsize=14, fontweight='bold')
for i, (val, pct) in enumerate(zip(values, percentages)):
    ax1.text(val + 50, i, f'{val:,} ({pct:.1f}%)', va='center', fontsize=10)

# Chart 2: Elegant pie chart
ax2 = axes[0, 1]
wedges, texts, autotexts = ax2.pie(values, labels=categories, autopct='%1.1f%%', 
                                    colors=colors, startangle=90, explode=[0.03]*5,
                                    shadow=True, textprops={'fontsize': 10})
ax2.set_title('Proportion of Cyberbullying Types', fontsize=14, fontweight='bold')

# Chart 3: Vertical bar chart
ax3 = axes[1, 0]
bars_v = sns.barplot(x=categories, y=percentages, palette=colors, ax=ax3)
ax3.set_xlabel('Category', fontsize=12)
ax3.set_ylabel('Percentage (%)', fontsize=12)
ax3.set_title('Percentage by Category', fontsize=14, fontweight='bold')
ax3.set_xticklabels(categories, rotation=45, ha='right')
for i, pct in enumerate(percentages):
    ax3.text(i, pct + 0.3, f'{pct:.1f}%', ha='center', fontsize=10)

# Chart 4: Donut chart
ax4 = axes[1, 1]
wedges2, texts2, autotexts2 = ax4.pie(values, labels=categories, autopct='%1.1f%%',
                                       colors=colors, startangle=90,
                                       wedgeprops=dict(width=0.5), textprops={'fontsize': 10})
ax4.set_title('Distribution (Donut Chart)', fontsize=14, fontweight='bold')
centre_circle = plt.Circle((0, 0), 0.25, fc='white')
ax4.add_artist(centre_circle)
ax4.text(0, 0, f'Total:\n{total:,}', ha='center', va='center', fontsize=12, fontweight='bold')

plt.suptitle('Cyberbullying Types Analysis', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
